```python
import numpy as np
from qiskit.quantum_info import Statevector

# Create the basic qubit states
zero = Statevector([1, 0])
one = Statevector([0, 1])

# Create an equal superposition
plus = Statevector([1/np.sqrt(2), 1/np.sqrt(2)])

print("State |0>:", zero.data)
print("State |1>:", one.data)
print("State |+>:", plus.data)
```

```python
# Create an invalid statevector
invalid = Statevector([1, 1])

# Check which states are normalized
print("|0> valid:", zero.is_valid())
print("|1> valid:", one.is_valid())
print("|+> valid:", plus.is_valid())
print("Invalid state valid:", invalid.is_valid())
```

```python
from qiskit.visualization import plot_histogram

# Set the random seed for repeatable results
plus.seed(42)

# Measure the state 1024 times
counts = plus.sample_counts(shots=1024)

print("Measurement counts:", counts)

# Show the measurement histogram
display(plot_histogram(counts))
```
