```python
# Experiment 03 — Inner Products and Tensor Products
```

```python
import numpy as np
from qiskit.quantum_info import Statevector

# Create the |0> and |1> states
zero = Statevector([1, 0])
one = Statevector([0, 1])

print("State |0>:", zero.data)
print("State |1>:", one.data)
```

```python
# Calculate the inner product of |0> and |1>
inner_01 = np.vdot(zero.data, one.data)

# Calculate the inner product of |0> with itself
inner_00 = np.vdot(zero.data, zero.data)

print("<0|1> =", inner_01)
print("<0|0> =", inner_00)
```

```python
# Combine |0> and |1> to form a two-qubit state
state_01 = zero.tensor(one)

# Combine |1> and |0> to form another state
state_10 = one.tensor(zero)

print("|0> tensor |1> =", state_01.data)
print("|1> tensor |0> =", state_10.data)
```

```python
# Calculate the same tensor product using NumPy
result = np.kron(zero.data, one.data)

print("NumPy tensor product:", result)

# Check whether both methods give the same result
print("Results match:", np.allclose(state_01.data, result))
```
