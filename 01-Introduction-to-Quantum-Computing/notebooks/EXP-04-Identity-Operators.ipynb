```python
# Experiment 04 — Identity Operators for One, Two and Three Qubits
```

```python
import numpy as np

# Create identity matrices for 1, 2 and 3 qubits
I1 = np.eye(2, dtype=int)
I2 = np.eye(4, dtype=int)
I3 = np.eye(8, dtype=int)

print("1-Qubit Identity Matrix:\n", I1)
print("\n2-Qubit Identity Matrix:\n", I2)
print("\n3-Qubit Identity Matrix:\n", I3)
```

```python
# Combine identity matrices using tensor products
I2_tensor = np.kron(I1, I1)
I3_tensor = np.kron(I2_tensor, I1)

print("2-Qubit Identity using Tensor Product:\n", I2_tensor)
print("\n3-Qubit Identity using Tensor Product:\n", I3_tensor)

print("\n2-Qubit matrices match:", np.array_equal(I2, I2_tensor))
print("3-Qubit matrices match:", np.array_equal(I3, I3_tensor))
```

```python
from qiskit.quantum_info import Statevector, Operator

# Create a single-qubit state
state1 = Statevector([1, 0])

# Apply the identity operator
result1 = state1.evolve(Operator(I1))

print("Original 1-Qubit State:", state1.data)
print("After Identity:", result1.data)

# Create a two-qubit state
state2 = Statevector([0, 1, 0, 0])
result2 = state2.evolve(Operator(I2))

print("\nOriginal 2-Qubit State:", state2.data)
print("After Identity:", result2.data)
```

```python
from qiskit import QuantumCircuit

# Create circuits for 1, 2 and 3 qubits
for n in [1, 2, 3]:
    qc = QuantumCircuit(n)

    # Apply identity gates to all qubits
    for i in range(n):
        qc.id(i)

    print(f"{n}-Qubit Identity Circuit:")
    display(qc.draw("mpl"))
```
