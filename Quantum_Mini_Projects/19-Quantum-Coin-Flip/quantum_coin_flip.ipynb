# Quantum Coin Flip

**Day 19 | 30-Day Quantum Computing Challenge**

## Aim:
To simulate a quantum coin flip using a Hadamard gate and study the probability of measuring 0 and 1.

## Tools:
Python, Qiskit, Qiskit Aer, Matplotlib

## Expected Result:
Approximately 50% zeros and 50% ones.

**Execution:** Run the cells from top to bottom. The notebook uses an ideal simulator unless explicitly stated otherwise.

```python
# Import the tools required to create and simulate a quantum circuit

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import matplotlib.pyplot as plt

# Number of times we will measure the circuit
shots = 1000
```

```python
# Create a circuit with one qubit and one classical bit
qc = QuantumCircuit(1, 1)

# Apply a Hadamard gate to create an equal superposition
qc.h(0)

# Measure the qubit and store the result in the classical bit
qc.measure(0, 0)

# Display the quantum circuit
print(qc.draw(output="text"))
```

```python
# Create the quantum circuit simulator
simulator = AerSimulator()

# Execute the circuit 1000 times
job = simulator.run(qc, shots=shots, seed_simulator=19)

# Retrieve the simulation results
result = job.result()
counts = result.get_counts(qc)

# Display the number of zeros and ones
print("Measurement results:", counts)
```

```python
# Count the occurrences of each measurement outcome
zeros = counts.get("0", 0)
ones = counts.get("1", 0)

# Convert the counts into percentages
prob_zero = zeros / shots * 100
prob_one = ones / shots * 100

print(f"Probability of 0: {prob_zero:.2f}%")
print(f"Probability of 1: {prob_one:.2f}%")
```

```python
# Plot the observed measurement probabilities
outcomes = ["0", "1"]
probabilities = [prob_zero, prob_one]

plt.figure(figsize=(7, 4))

plt.bar(outcomes, probabilities, color=["#4676C9", "#8064C9"])

# Show the theoretical probability for comparison
plt.axhline(50, color="black", linestyle="--",
            label="Theoretical probability (50%)")

plt.title("Quantum Coin Flip — Measurement Results")
plt.xlabel("Measurement Outcome")
plt.ylabel("Observed Probability (%)")
plt.ylim(0, 100)
plt.legend()
plt.tight_layout()

# Save the graph for GitHub documentation
plt.savefig("results.png", dpi=200)
plt.show()
```

```python
## Observation

The Hadamard gate transforms the initial state |0⟩ into an
equal superposition of |0⟩ and |1⟩.

After 1,000 measurements, both outcomes occur with
approximately equal frequency.

The observed percentages may differ slightly from 50%
because of statistical fluctuations.

## Conclusion

A quantum coin flip was successfully simulated using
a single-qubit Hadamard circuit.

The experiment demonstrates superposition, quantum
measurement, and the probabilistic nature of quantum outcomes.

The results were obtained using a quantum circuit simulator,
not physical quantum hardware.
```
