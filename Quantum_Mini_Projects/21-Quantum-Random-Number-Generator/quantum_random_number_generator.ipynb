## Aim:
To generate a binary sequence from simulated quantum measurements and analyze its bit distribution.

## Tools:
Python, Qiskit, Qiskit Aer, Matplotlib

## Expected Result:
A binary sequence with approximately equal numbers of zeros and ones, subject to sampling fluctuations.

**Execution:** Run the cells from top to bottom. The notebook uses an ideal simulator unless explicitly stated otherwise.

## 1. Idea
A qubit initialized to $|0\rangle$ and acted on by $H$ yields equal theoretical probabilities for outcomes 0 and 1. Repeated measurements create a bit sequence.

```python
# Experiment step: execute this cell and inspect its output.
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

number_of_bits = 100  # change this to 10, 1000, ...
seed = 2026  # fixed seed for reproducibility
assert isinstance(number_of_bits, int) and number_of_bits > 0
rng = np.random.default_rng(seed)
bits = rng.integers(0, 2, size=number_of_bits)
bit_string = "".join(map(str, bits))
print("Generated bit sequence:", bit_string)
```

## 2. Frequency analysis
The expected fraction is 50% for each outcome, but finite samples vary.

```python
# Experiment step: execute this cell and inspect its output.
zeros = int(np.count_nonzero(bits == 0))
ones = int(np.count_nonzero(bits == 1))
print(f"Total: {number_of_bits} | 0s: {zeros} ({zeros/number_of_bits:.2%}) | 1s: {ones} ({ones/number_of_bits:.2%})")
print("Absolute imbalance:", abs(zeros - ones))
```

## 3. Visualize the experiment
A frequency histogram and a running proportion plot reveal the sample fluctuations.

```python
# Experiment step: execute this cell and inspect its output.
fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].bar(["0", "1"], [zeros, ones], color=["#5166b6", "#b38acb"])
axes[0].axhline(number_of_bits/2, color="#dd8b36", ls="--", label="Expected 50%")
axes[0].set(title=f"Bit frequency ({number_of_bits} bits)", ylabel="Count")
axes[0].legend()
axes[1].step(np.arange(1, number_of_bits+1), np.cumsum(bits == 1)/np.arange(1, number_of_bits+1), where="mid", color="#5166b6")
axes[1].axhline(0.5, color="#dd8b36", ls="--")
axes[1].set(xlabel="Bits generated", ylabel="Fraction of 1s", title="Running fraction of 1s", ylim=(0,1))
fig.tight_layout()
fig.savefig("results.png", dpi=180, bbox_inches="tight")
plt.show()
```

## 4. Optional: simulate a real quantum circuit in Qiskit Aer
Install `qiskit qiskit-aer` first. **Aer is a classical simulator** of a quantum circuit; this does not generate hardware-certified quantum entropy.

```python
# Experiment step: execute this cell and inspect its output.
try:
    from qiskit import QuantumCircuit, transpile
    from qiskit_aer import AerSimulator
    qc = QuantumCircuit(1, 1)
    qc.h(0)
    qc.measure(0, 0)
    print(qc.draw(output="text"))
    backend = AerSimulator()
    result = backend.run(transpile(qc, backend), shots=number_of_bits, memory=True).result()
    qiskit_bits = result.get_memory()
    print("Qiskit Aer sequence:", "".join(qiskit_bits))
    print("Qiskit Aer counts:", result.get_counts())
except ImportError:
    print("Optional Qiskit section skipped. Install: pip install qiskit qiskit-aer")
```

## 5. Interpretation and limits
- A balanced histogram is **not proof of randomness** or cryptographic suitability.
- NumPy uses pseudorandom sampling of the theoretical distribution.
- Qiskit Aer simulates gate and measurement behavior; it is not a physical QRNG.
- For a real QRNG, entropy source characterization, bias correction, and independent tests are needed.
